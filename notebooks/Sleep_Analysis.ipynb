# Circadia — Sleep Analysis

Thin analysis notebook. All pipeline logic lives in the `condor/` library.

**Quick start**
1. Set `FILE` and `DEVICE` below.
2. Run all cells.

```python
import sys, pathlib

# Repo root — works regardless of where Jupyter's cwd is
REPO_ROOT = pathlib.Path('__file__').resolve().parent.parent

# Ensure condor is importable if not installed via pip
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

print(f"Repo root: {REPO_ROOT}")
```

```python
# ── Configuration ─────────────────────────────────────────────────────────────
FILE   = REPO_ROOT / "notebooks" / "input1.txt"
DEVICE = "acttrust"   # 'acttrust' | 'actiwatch'
```

```python
from condor_pipeline.pipeline import SleepPipeline

pipe    = SleepPipeline(FILE, device=DEVICE)
results = pipe.run()
```

```python
# Nightly sleep statistics
results.nights
```

```python
# Timestamp consistency issues (empty = none detected)
results.issues
```

```python
# Final actogram
results.plot()
```
